# Sección 2 — ¿Cómo cambia el pangenoma al añadir genomas?

Para cada orden aleatorio de incorporación de genomas calcularemos:

- **Pangenoma acumulado:** familias presentes en al menos uno de los genomas incluidos.
- **Core acumulado:** familias presentes en todos los genomas incluidos.

Usaremos 100 permutaciones de la matríz de presencia y ausencia (objeto `presence`). Con 250 × 16 datos, el cálculo es muy ligero.

````python
# Ejercicio 2: ¿Cómo cambia el pangenoma al agregar genomas?

# Definiendo una función que va a muestrear genomas. Esta función incluye 2 parámetros:
# binary_matrix: La matríz de presencia / ausencia de familias de genes en los genomas (Objeto presence).
# order: el órden en que se agregaran los 16 genomas. Éste ordén será aleatorio y veremos el efecto que tiene en el pangenoma y el genoma core.
def accumulation_for_order(binary_matrix, order):

    # Aquí definimos listas en las cuales un for loop agregaran las estimaciones de familias presentes en el pangenoma (observadas en al menos 1 genoma del submuestreo) 
    # y presentes en el core (presentes en todos los genomas del submuestreo)
    pan_sizes = []
    core_sizes = []

    # En esta lista guardaremos los genomas seleccionados
    selected = []

    # El for loop nos permitirá realizar las iterar cada vez que agreguemos un nuevo genoma
    for genome in order:

        # Agregando un genoma seleccionado aleatoriamente. Este genoma será adicionado al objeto "selected", por lo que, en cada iteración del for loop irán aumentando el número de genomas
        print('Genoma: ',genome)
        selected.append(genome)
        print('Número de genomas adicionados hasta el momento: ', len(selected))
        
        # Aquí realizaremos filtraremos la matríz de presencias / ausencias a los genomas que llevamos muestrados
        sub = binary_matrix[selected]

        # Aquí estimamos el número de familias presentes en al menos un genoma del submuestreo:
        pan_sizes.append((sub.sum(axis=1) _____________ ).sum())

        # Número de familias presentes en todos los genomas seleccionados hasta el momento (Pista: como sabemos cuantos genomas llevamos seleccionados hasta ahora?):
        core_sizes.append((sub.sum(axis=1) == _____________).sum())

    # Terminando los 16 genomas la función devolverá los tamaños de pangenoma y cores estimados al ir agregando un genoma aleatorio a la vez
    return np.array(pan_sizes), np.array(core_sizes)

# Correremos múltiples veces la función que hemos creado!
# Recuerda que nuestra función estima tamaño del pangenoma y core cada vez que adiciona un genoma aleatorio nuevo
# Adicionando un valor de semilla aleatoria para que nuestros resultados sean replicables
rng = np.random.default_rng(1234)

# Aquí incluimos las 100 iteraciones en las que correremos nuestra función
reps = 100

# Creando 2 listas nuevas. En ellas agregaremos los pangenomas y cores estimados en cada uno de los 16 calculos de las 100 iteraciones 
pan_all, core_all = [], []

# Guardando nombres de los genomas
columns = list(presence.columns)

# En este segundo for loop repetiremos nuestra función 100 veces
for i in range(reps):
    print("Iteración :", i)
    
    # Ojo! Aquí cambiamos el orden de los genomas de forma aleatoria para cada una de las 100 iteraciones!
    order = list(rng.permutation(columns))
    print("Ordén: ", order)

    # Aquí corremos nuestra función en la matríz de presencia ausencia con cada órden nuevo
    pan, core = accumulation_for_order(presence, order)

    # Y aquí añadimos los 16 resultados de cada corrida a la lista global de las 100 iteraciones
    pan_all.append(pan)
    core_all.append(core)

# Estos ultimos comandos unen las 2 listas de 100 listas que contienen cada una 16 estimaciones de pangenoma y core en 2 listas de 1600 valores.
pan_all = np.vstack(pan_all)
core_all = np.vstack(core_all)

# Esta cell imprime mucho texto. Puedes limpiarlo en Edit > Clear cell output.
````

# Sección 3 — Similitud de contenido génico

La similitud de Jaccard compara los genes compartidos entre dos genomas respecto al total observado en ambos. Usaremos la **distancia de Jaccard** para agrupar genomas por contenido génico.

**Advertencia:** un dendrograma de presencia/ausencia **no es automáticamente una filogenia**. Ganancias, pérdidas, transferencia horizontal y errores de anotación pueden modificar el contenido génico.

````python
# Matriz de distancia Jaccard entre genomas
# pdist espera observaciones en filas, por eso transponemos (método ".T.values").
# Recuerda que pdist viene de la paquetería "scipy.spatial.distance"
D = pdist(presence.T.values, metric='jaccard')
D_square = pd.DataFrame(squareform(D), index=presence.columns, columns=presence.columns)
display(D_square.round(3))

# Una vez que tenemos las distancias podemos realizar el dendograma agrupando los genomas por similaridad de forma jerarquica
Z = linkage(D, method='average')
````

````python
# Figura 3: dendrograma de similitud por contenido génico
plt.figure(figsize=(10,6))
dendrogram(Z, labels=presence.columns, leaf_rotation=90)
plt.ylabel('Jaccard distance')
plt.title('Gene-content clustering — teaching matrix')
plt.tight_layout()
plt.savefig('../results/Fig3_gene_content_dendrogram.png', dpi=180)
plt.show()
````
