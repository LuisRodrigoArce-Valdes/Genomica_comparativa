# Sesión 2 — ¿Cómo cambia el pangenoma al añadir genomas?

Para cada orden aleatorio de incorporación de genomas calcularemos:

- **Pangenoma acumulado:** familias presentes en al menos uno de los genomas incluidos.
- **Core acumulado:** familias presentes en todos los genomas incluidos.

Usaremos 100 permutaciones. Con 250 × 16 datos, el cálculo es muy ligero.

```python
# Figura 2A: acumulación del pangenoma
x = np.arange(1, presence.shape[1]+1)
plt.figure(figsize=(8,5))
plt.plot(x, pan_all.mean(axis=0), marker='o')
plt.fill_between(x, pan_all.min(axis=0), pan_all.max(axis=0), alpha=0.15)
plt.xlabel('Number of genomes')
plt.ylabel('Cumulative gene families')
plt.title('Pangenome accumulation — teaching matrix')
plt.tight_layout()
plt.savefig('Fig2A_pangenome_accumulation.png', dpi=180)
plt.show()
```

```python
# Figura 2B: reducción del core
plt.figure(figsize=(8,5))
plt.plot(x, core_all.mean(axis=0), marker='o')
plt.fill_between(x, core_all.min(axis=0), core_all.max(axis=0), alpha=0.15)
plt.xlabel('Number of genomes')
plt.ylabel('Core gene families')
plt.title('Core-genome accumulation — teaching matrix')
plt.tight_layout()
plt.savefig('Fig2B_core_accumulation.png', dpi=180)
plt.show()
```

### Preguntas de interpretación

# Sesión 3 — Similitud de contenido génico
